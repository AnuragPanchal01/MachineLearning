```python
df = pd.read_csv('/Users/anuragpanchal/Documents/MachineLearning/Logistic-Regression-Bank/bank.csv',delimiter=';')
```

```python
cat_cols = df.select_dtypes(['object']).columns
from sklearn.preprocessing import LabelEncoder
for i in cat_cols:
    df[i] = LabelEncoder().fit_transform(df[i])
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
x = df.iloc[:,:-1]
x
```

```python
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.feature_selection import RFE
model  = LogisticRegression()
```

```python
rfe = RFE(model)
```

```python
rfe.fit(x,y)
```

```python
rfe.support_
```

```python
selected_index = [i for i ,data in enumerate(rfe.support_) if data==True]
```

```python
selected_index
```

```python
cols = x.columns
```

```python
selected_columns = [cols[i] for i in selected_index]
```

```python
selected_columns
```

```python
X = x.loc[:,selected_columns]
```

```python
X
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```
