```python
df.drop(columns=['Player.x','Player_ID'],inplace=True)
```

```python
for i in cat_cols:
    df[i] = LabelEncoder().fit_transform(df[i])
```

```python
df.info()
```

```python
df['Play'].value_counts(normalize=True)*100
```

```python
x = df.drop(columns=
            'Play')
y = df['Play']
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=200,stratify=y)
```

```python
sc = StandardScaler()
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
```

```python
model = LogisticRegression()
model.fit(x_train_scaled,y_train)
```

```python
y1 = model.predict(x_test_scaled)
```

```python
accuracy_score(y_test,y1)
```

```python
roc_auc_score(y_test,y1)
```

```python
pca = PCA(n_components=0.90)
```

```python
x_train_pca = pca.fit_transform(x_train_scaled)
x_test_pca = pca.transform(x_test_scaled)
```

```python
x_train_pca.shape
```

```python
model2 = LogisticRegression()
```

```python
model2.fit(x_train_pca,y_train)
y2 = model2.predict(x_test_pca)
```

```python
accuracy_score(y_test,y2)
```

```python
roc_auc_score(y_test,y2)
```
